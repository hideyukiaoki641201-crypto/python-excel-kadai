```python
# ① 必要なライブラリを読み込む
import pandas as pd

# ② 社員の売上データを作成する
data = {
    "日付": [
        "2023-05-17",
        "2023-05-18",
        "2023-05-19",
        "2023-05-20",
        "2023-05-21"
    ],
    "社員名": ["山田", "佐藤", "鈴木", "田中", "高橋"],
    "売上": [100, 200, 150, 300, 250],
    "部門": ["メーカー", "代理店", "メーカー", "商社", "代理店"]
}

# ③ 辞書データをDataFrame（表形式のデータ）に変換する
df = pd.DataFrame(data)

# ④ 全社員の平均売上を計算する
average_sales = df["売上"].mean()

# ⑤ 平均売上の列を追加する
df["平均売上"] = average_sales

# ⑥ 業績ランクを判定する関数を定義する
def performance(sales):
    if sales >= average_sales + 50:
        return "A"
    elif sales >= average_sales:
        return "B"
    else:
        return "C"

# ⑦ 各社員の売上に関数を適用し、業績ランク列を作成する
df["業績ランク"] = df["売上"].apply(performance)

# ⑧ Excelファイルに書き出す
df.to_excel("業績.xlsx", index=False)

# ⑨ 結果を画面に表示する
print(df)

print("業績.xlsxを作成しました")
```
