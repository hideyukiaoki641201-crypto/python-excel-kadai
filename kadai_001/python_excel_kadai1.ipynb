```python

# Step 1. 必要なライブラリを読み込む
import openpyxl
from openpyxl.styles import Font
from datetime import datetime, timedelta, timezone
from google.colab import files

# Step 2. 現在の日付を日本時間で取得する
today = datetime.now(timezone(timedelta(hours=9)))

# 請求書内の日付をYYYY/MM/DD形式に変換する
date_str = today.strftime("%Y/%m/%d")

# ファイル名用の日付をYYYYMMDD形式に変換する
file_date = today.strftime("%Y%m%d")

# Step 3. Excelファイルを新規作成する
wb = openpyxl.Workbook()

# 操作対象のワークシートを取得する
ws = wb.active

# Step 4. 請求書のタイトルと会社情報を入力する
ws["B2"] = "請求書"
ws["B4"] = "株式会社ABC"
ws["B5"] = "〒101-0022 東京都千代田区神田練塀町300"
ws["B6"] = "TEL:03-1234-5678 FAX:03-1234-5678"
ws["B7"] = "担当者名:鈴木一郎 様"

# Step 5. 請求書番号と現在の日付を入力する
ws["F4"] = "No."
ws["G4"] = "0001"
ws["F5"] = "日付"
ws["G5"] = date_str

# Step 6. 商品明細の見出しを入力する
ws["B10"] = "商品名"
ws["C10"] = "数量"
ws["D10"] = "単価"
ws["E10"] = "金額"

# Step 7. 商品データをリストに格納する
# データの順番は商品名、数量、単価
products = [
    ["商品A", 2, 10000],
    ["商品B", 1, 15000]
]

# Step 8. 合計金額の初期値を0に設定する
total = 0

# Step 9. for文で商品データを繰り返し入力する
# enumerateで行番号を11から取得する
for i, product in enumerate(products, start=11):

    # リストから商品名、数量、単価を取り出す
    name = product[0]
    quantity = product[1]
    price = product[2]

    # 数量と単価を掛けて金額を計算する
    amount = quantity * price

    # Excelの各セルにデータを入力する
    ws.cell(row=i, column=2, value=name)
    ws.cell(row=i, column=3, value=quantity)
    ws.cell(row=i, column=4, value=price)
    ws.cell(row=i, column=5, value=amount)

    # 商品ごとの金額を合計する
    total += amount

# Step 10. 消費税と税込合計を計算する
tax = int(total * 0.1)
grand_total = total + tax

# Step 11. 合計金額をExcelに入力する
ws["E13"] = total

ws["B15"] = "合計"
ws["E15"] = total

ws["B16"] = "消費税"
ws["E16"] = tax

ws["B17"] = "税込合計"
ws["E17"] = grand_total

# Step 12. タイトルと見出しを太字にする
ws["B2"].font = Font(bold=True)

for cell in ws["B10:E10"][0]:
    cell.font = Font(bold=True)

# Step 13. 現在の日付を含むファイル名を作成する
filename = f"請求書_{file_date}.xlsx"

# Step 14. Excelファイルを保存する
wb.save(filename)

# Step 15. 作成結果を表示する
print(f"{filename}を作成しました")
print(f"合計金額：{total}円")
print(f"消費税：{tax}円")
print(f"税込合計：{grand_total}円")

# Step 16. Excelファイルをダウンロードする
files.download(filename)

```
