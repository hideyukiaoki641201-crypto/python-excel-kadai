```python
# Step 1. 必要なライブラリを読み込む
import openpyxl
from datetime import datetime

# Step 2. 現在の日付を取得する
today = datetime.now()

# 日付をYYYY/MM/DD形式に変換する
date_str = today.strftime("%Y/%m/%d")

# Step 3. Excelファイルを新規作成する
wb = openpyxl.Workbook()

# 操作対象のワークシートを取得する
ws = wb.active

# Step 4. セルにデータを書き込む
ws["B2"] = "請求書"
ws["B4"] = "株式会社ABC"

# Step 5. ファイル名を作成する
# 現在の日付をYYYYMMDD形式に変換する
file_date = today.strftime("%Y%m%d")
filename = f"請求書_{file_date}.xlsx"

# Step 6. Excelファイルを保存する
wb.save(filename)

# Step 7. 処理の完了を表示する
print(f"{filename}を作成しました")

```
